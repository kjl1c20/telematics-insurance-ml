```python
import numpy as np

numeric_cols = demo.select_dtypes(include="number").columns
rate_cols = [c for c in numeric_cols if c.startswith("Accel.") or c.startswith("Brake.") or "turn.intensity" in c.lower()]

# Accel/Brake/turn intensity columns are "per 1000 miles" rates that blow up to
# extreme values for low-mileage policies, which squashes their histograms flat.
# log1p compresses that long tail so the real shape is visible; it's safe here
# since these columns are all >= 0 (log1p(0) = 0).
demo_hist = demo[numeric_cols].copy()
demo_hist[rate_cols] = demo_hist[rate_cols].apply(np.log1p)

axes = demo_hist.hist(
    bins=30,
    figsize=(20, 24),
    layout=(-1, 4),
    edgecolor="black"
)

for ax, col in zip(axes.flatten(), demo_hist.columns):
    if col in rate_cols:
        ax.set_title(f"log1p({col})")

plt.tight_layout()
plt.show()
```

The histogram above shows that there is clearly an outlier in acceleration related columns that I need to investigate further. The above plot was adjusted to use log scale to look at the rates.

```python
ncols = 4
nrows = -(-len(numeric_cols) // ncols)  # ceil division
fig, axs = plt.subplots(nrows, ncols, figsize=(20, 3.5 * nrows))
axs = axs.flatten()

# with 100,000 rows, outlier markers overlap and render as a solid black bar;
# shrinking and fading them turns that into a readable density gradient instead
flierprops = dict(marker="o", markersize=3, markerfacecolor="gray", alpha=0.4, markeredgecolor="none")

for ax, col in zip(axs, numeric_cols):
    # reuse the same log1p-transformed data as the histogram above for rate_cols,
    # so outliers in Accel/Brake/turn intensity are visible instead of squashing the box flat
    data = demo_hist[col] if col in rate_cols else demo[col]
    ax.boxplot(data, orientation="vertical", flierprops=flierprops)
    ax.set_title(f"log1p({col})" if col in rate_cols else col, fontsize=9)

for ax in axs[len(numeric_cols):]:
    ax.set_visible(False)

plt.tight_layout()
plt.show()
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

# convert the "Territory" column to a categorical type for VIF calculation
demo["Territory"] = demo["Territory"].astype("category")
# drop the target variables from the feature set for VIF calculation
X = demo.drop(columns=["NB_Claim", "AMT_Claim", "Duration"])

cat_columns = X.select_dtypes(include=["object", "category"]).columns.tolist()
cont_columns = X.select_dtypes(include=["number"]).columns.tolist()
print(f"Categorical columns: {cat_columns}")
print(f"Continuous columns: {cont_columns}")
print(f"Sanity Check - Number of columns: {len(cat_columns)+len(cont_columns)}")

# Calculate VIF for each feature
# statmodels post-0.15 has standardized=True so we don't need to add a constant before the calculation
vif_data = pd.DataFrame()
vif_data['Cont_Feature'] = cont_columns
vif_data['VIF'] = [variance_inflation_factor(X[cont_columns].values, i) for i in range(X[cont_columns].shape[1])]


print(vif_data)
```

Based on the results above, there are redundant columns like that could be derived from other columns. We should remove "Pct.drive.wkday", "Pct.drive.wkend" as they can be derived using individual day data. We can drop "Pct.drive.sun" as well as it can be derived from the other six days. We won't drop the telematics columns. Those telematics columns are nested features. Might use ridge penalty to address it in the ML later.

```python
vif_data = pd.DataFrame()
cont_columns = [x for x in cont_columns if x not in ["Pct.drive.sun", "Pct.drive.wkday", "Pct.drive.wkend"]]
vif_data['Cont_Feature'] = cont_columns
vif_data['VIF'] = [variance_inflation_factor(X[cont_columns].values, i) for i in range(X[cont_columns].shape[1])]


print(vif_data)
```

### Collinearity across categorical and continuous features (GVIF)

```python
design_df = pd.get_dummies(X[cont_columns + cat_columns], columns=cat_columns, drop_first=True)

# GVIF treats a multi-dummy factor as one term, so map each original feature
# to all of its design-matrix column(s) instead of scoring dummies individually.
term_columns = {col: [col] for col in cont_columns}
for cat in cat_columns:
    term_columns[cat] = [c for c in design_df.columns if c.startswith(f"{cat}_")]

R = design_df.corr().values
col_index = {col: i for i, col in enumerate(design_df.columns)}
det_R = np.linalg.det(R)

# GVIF_j = det(R_jj) * det(R_(-j)(-j)) / det(R), where R_jj is the correlation
# submatrix of term j's own column(s) and R_(-j)(-j) is that of every other
# term's columns (Fox & Monette, 1992).
gvif_rows = []
for term, cols in term_columns.items():
    idx = [col_index[c] for c in cols]
    other_idx = [i for i in range(len(design_df.columns)) if i not in idx]

    det_term = np.linalg.det(R[np.ix_(idx, idx)])
    det_other = np.linalg.det(R[np.ix_(other_idx, other_idx)])

    gvif = (det_term * det_other) / det_R
    df = len(idx)
    gvif_adj = gvif ** (1 / (2 * df))  # comparable to sqrt(VIF)

    gvif_rows.append({"Feature": term, "df": df, "GVIF": gvif, "GVIF^(1/(2*df))": gvif_adj})

gvif_data = pd.DataFrame(gvif_rows).sort_values("GVIF^(1/(2*df))", ascending=False)
print(gvif_data)
```
