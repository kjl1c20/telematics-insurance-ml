```python
import pandas as pd
import matplotlib.pyplot as plt
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
demo = pd.read_csv("telematics_syn-032021.csv")
```

```python
print(f"Number of columns: {len(demo.columns)}")
demo.describe(include='all')
```

### Quick summary
The above results showed that this sample data is very likely to be taken in a 366 days year so when we convert the duration unit from days to years we will divide by 366. Nothing odd for categorical columns except for the car age column. The negative car age is reasonable as the car age baseline is based on model year. Negative values means buying a car model with model year ahead of calendar year. 
   
Numerical variables are in sensible range too. The total miles driven in a year is quite sensible. All the acceleration metrics looks to be sensible too.
