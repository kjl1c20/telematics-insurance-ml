# Pandas EDA
1. Get table information like not null count and data type using df.info()
2. Get high level stats using df.describe()
3. Inspect sample data using df.head()
4. df.isnull().sum(): Returns the number of missing values per column.
5. Use plt.pie() to visualise target variable distribution.


source: https://www.geeksforgeeks.org/data-analysis/data-preprocessing-machine-learning-python/

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
demo = pd.read_csv("telematics_syn-032021.csv")
```

```python
print(f"Number of columns: {len(demo.columns)}")
demo.describe(include='all')
```

```python
demo.head()
```

```python
demo.info()
```

```python
fig, axs = plt.subplots(len(demo.columns), 1, figsize=(7, 18), dpi=95)
for i, col in enumerate(demo.columns):
    axs[i].boxplot(demo[col], vert=False)
    axs[i].set_ylabel(col)
plt.tight_layout()
plt.show()
```
